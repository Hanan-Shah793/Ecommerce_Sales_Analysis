Ecommerce_Sale_Dataset:

```python
import pandas as pd
import numpy as np

```

```python
#importing_Ecommerce_Sales_Dataset_For_Cleaning_&_Visualizing :

var = pd.read_csv("ecommerce_sales_dataset_v2.csv")
df = pd.DataFrame(var)
df
```

```python
#using_These_Functions_For_The_Basic_Knowledge_About_Data :


df.shape
print()
df.columns
print()
df.info()
print()
df.describe()
```

```python
#if any date_time column data_type is "object" so for its  correction  we use this method :

df["OrderDate"] = pd.to_datetime(df["OrderDate"])
```

```python
#Using this Function for checking null values in dataset :

df.isnull().sum()
```

```python
#Using this Function for checking duplicates in dataset :

df.duplicated()
```

```python
# Creating a "Total_revenue" column with the help of "price" & "Quantity" column:

df["Total_revenue"] = df["Price"] * df["Quantity"]
df
```

```python
#For checking Best Selling Product by this way :

Best_Selling_product = df.groupby("ProductName")["Quantity"].sum()
Best_Selling_product
```

```python
# Top "5" Selling Products :

Best_Selling_product = df.groupby("ProductName")["Quantity"].sum().sort_values(ascending=False)
Best_Selling_product.head(5)
```

```python
#visulization form of Best Selling Product :

import matplotlib.pyplot as plt 
import seaborn as sns

Best_Selling_product.plot(kind="bar" , edgecolor = "black" ,  linewidth = 3 , width = 0.8 , colormap="plasma" , label = "Best_Selling_product")

plt.title("Best_Selling_product" , fontsize = 14)
plt.xlabel("Product_Name" , fontsize = 14)
plt.ylabel("Quantity_Sold" , fontsize = 14)

plt.legend()
plt.show()
```

```python
#For Checking Best Selling Category :

Best_Selling_category = df.groupby("ProductCategory")["Quantity"].sum()
Best_Selling_category
```

```python
#Visualization Form of Best Selling Category see this for your better analyization :

import matplotlib.pyplot as plt
import seaborn as sns

Best_Selling_category.plot(kind="bar" , edgecolor = "black" , linewidth = 3 , colormap="viridis" , label = "Best_Selling_category")



plt.title("Best_Selling_Category" , fontsize = 15)
plt.xlabel("category" , fontsize = 15)
plt.ylabel("Quantity_Sold" , fontsize = 15)
plt.xticks(rotation = 0)

plt.legend()
plt.show()
```

```python
#For checking Different Cities Sales :

city_sales = df.groupby("City")["Total_revenue"].sum()
city_sales
```

```python
#Top "3" Best Cities Sales :

Top_city_sale = city_sales = df.groupby("City")["Total_revenue"].sum().sort_values(ascending=False)
Top_city_sale.head(3)
```

```python
# visualization form of city_sales see this for your better analyization :


import matplotlib.pyplot as plt
import seaborn as sns

city_sales.plot(kind="bar" , edgecolor = "black" , linewidth = 3 , colormap="cividis" , label = "City_Sales")


plt.title("city_sales" , fontsize = 16)
plt.xlabel("Cities" , fontsize = 15)
plt.ylabel("Revenue" , fontsize = 15)

plt.legend()
plt.xticks(rotation = 0)
plt.show()
```

```python
#For checking the average of the order values :

avg_order_value = df["Total_revenue"].sum()
avg_order_id = df["OrderID"].nunique()
avg = avg_order_value / avg_order_id
avg_order_value 
print()
avg_order_id
print()
avg
```

```python
#For checking Daily_Sales by this way :

daily_sale = df.groupby(["OrderDate"])["Total_revenue"].sum()
daily_sale
```

```python
# visualization form of the Daily_Sales see this for your better analyization : 

import matplotlib.pyplot as plt 
import seaborn as sns

daily_sale.plot(kind="line" , label = "Daily_Sale" , color = "m")


plt.title("Daily_Sales" , fontsize = 20)
plt.xlabel("Date" , fontsize = 15)
plt.ylabel("Total_Sales" , fontsize  = 15)

plt.legend()
plt.show()
```
